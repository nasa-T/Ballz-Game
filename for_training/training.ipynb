```python
import gymnasium as gym
import numpy as np
from ai_trainer import *
```

```python
gym.register("gymnasium_env/Ballz-v0", entry_point=BallzEnv, max_episode_steps=100)
```

```python
env = gym.make("gymnasium_env/Ballz-v0", width=700, height=950, nblocks=7, render_mode="human")
```

```python
env.observation_space
```

```python
agent = BallzAgent(env)
```

```python
rewards = agent.train(10000, live_plot=True)
```

```python
obs_list = []
for i in range(5):
    obs, _ = env.reset()
    for j in range(i):
        obs = env.step(0)[0]
    obs_list.append(obs)
    o_tens = agent.forward(obs)
    print(agent.actor(o_tens).numpy())
    print(agent.critic(o_tens).numpy())
```

```python
rewards = agent.train(20, live_plot=True)
```

```python
for obs in obs_list:
    o_tens = agent.forward(obs)
    print(agent.actor(o_tens).numpy())
    print(agent.critic(o_tens).numpy())
```
