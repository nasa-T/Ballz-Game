```python
import gymnasium as gym
import numpy as np
from ai_trainer import *
import datetime
%load_ext tensorboard
```

```python
gym.register("gymnasium_env/Ballz-v0", entry_point=BallzEnv, max_episode_steps=100)
```

```python
# env = gym.make("gymnasium_env/Ballz-v0", width=700, height=950, nblocks=7, render_mode=None)
env = gym.make_vec("gymnasium_env/Ballz-v0", num_envs=2, width=700, height=950, nblocks=7, render_mode="human", vectorization_mode="async")
env.set_attr("shared_memory",False)
```
