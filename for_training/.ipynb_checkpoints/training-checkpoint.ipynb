```python
import gymnasium as gym
import numpy as np
from ai_trainer import *
```

```python
gym.register("gymnasium_env/Ballz-v0", entry_point=BallzEnv, max_episode_steps=100)
```

```python
env = gym.make("gymnasium_env/Ballz-v0", width=700, height=950, nblocks=7, render_mode='human')
```

```python
env.observation_space
```

```python
from gymnasium.utils.env_checker import check_env
```

```python
check_env(env.unwrapped)
print("Environment passes all checks!")
```

```python
agent = BallzAgent(env)
```

```python
obs1, _ = env.reset()
```

```python
state = agent.forward(obs1)
```

```python
rewards = agent.train(10000)
```

```python
rewards
```

```python
import matplotlib.pyplot as plt
plt.plot(rewards)
plt.xlabel('epoch')
plt.ylabel('reward')
```
